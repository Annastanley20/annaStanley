```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import joblib
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from google.colab import files

# Load dataset and clean column names
df = pd.read_csv("healthcare_employability_dataset.csv")
df.columns = df.columns.str.strip()

# Select structural features
numerical_features = [
    'Experience_Years', 'Number_of_Skills', 'Employability_Score',
    'Communication_Level', 'Leadership_Level', 'Teamwork_Level',
    'Problem_Solving_Level', 'Digital_Literacy_Level'
]
categorical_features = ['Education_Level', 'Job_Level']
X = df[numerical_features + categorical_features].copy()

# Execute preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
    ]
)
X_processed = preprocessor.fit_transform(X)

# Execute Principal Component Analysis
pca = PCA(n_components=3, random_state=42)
X_pca = pca.fit_transform(X_processed)

# Train K-Means++ Model
kmeans = KMeans(n_clusters=4, init='k-means++', max_iter=300, n_init=10, random_state=42)
cluster_labels = kmeans.fit_predict(X_pca)

# Calculate metrics dynamically
calculated_silhouette = silhouette_score(X_pca, cluster_labels)
total_variance = np.sum(pca.explained_variance_ratio_) * 100

# Clean console output for screenshotting
print(f"Total Explained Variance by PCA: {total_variance:.2f}%")
print(f"Calculated Silhouette Score: {calculated_silhouette:.4f}")

# Save model binaries
joblib.dump(preprocessor, 'employability_preprocessor.pkl')
joblib.dump(pca, 'employability_pca_model.pkl')
joblib.dump(kmeans, 'employability_kmeans_model.pkl')

# Trigger browser file downloads
files.download('employability_preprocessor.pkl')
files.download('employability_pca_model.pkl')
files.download('employability_kmeans_model.pkl')
```
